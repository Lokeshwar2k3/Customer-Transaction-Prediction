```python
from sklearn.model_selection import train_test_split

X_fit, X_val, y_fit, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

print("Training data:", X_fit.shape)
print("Validation data:", X_val.shape)
```

```python
from sklearn.base import clone

# Clone existing models
xgb_tuned = clone(xgb_model)
lgbm_tuned = clone(lgbm_model)

# Train on the smaller training set
xgb_tuned.fit(X_fit, y_fit)
lgbm_tuned.fit(X_fit, y_fit)

print("XGBoost and LightGBM retraining completed.")
```

```python
import numpy as np
from sklearn.metrics import precision_recall_curve

def find_best_threshold(model, X_val, y_val):
    probabilities = model.predict_proba(X_val)[:, 1]

    precision, recall, thresholds = precision_recall_curve(
        y_val, probabilities
    )

    f1_scores = (
        2 * precision[:-1] * recall[:-1] /
        (precision[:-1] + recall[:-1] + 1e-10)
    )

    best_index = np.argmax(f1_scores)

    return thresholds[best_index], f1_scores[best_index]


xgb_threshold, xgb_val_f1 = find_best_threshold(
    xgb_tuned, X_val, y_val
)

lgbm_threshold, lgbm_val_f1 = find_best_threshold(
    lgbm_tuned, X_val, y_val
)

print("XGBoost best threshold:", round(xgb_threshold, 4))
print("XGBoost validation F1:", round(xgb_val_f1, 4))

print("LightGBM best threshold:", round(lgbm_threshold, 4))
print("LightGBM validation F1:", round(lgbm_val_f1, 4))
```

```python
# Retrain on complete training data
xgb_tuned.fit(X_train, y_train)
lgbm_tuned.fit(X_train, y_train)

# Get test probabilities
xgb_test_prob = xgb_tuned.predict_proba(X_test)[:, 1]
lgbm_test_prob = lgbm_tuned.predict_proba(X_test)[:, 1]

# Apply optimized thresholds
xgb_test_pred = (xgb_test_prob >= xgb_threshold).astype(int)
lgbm_test_pred = (lgbm_test_prob >= lgbm_threshold).astype(int)

from sklearn.metrics import classification_report

print("XGBOOST - OPTIMIZED THRESHOLD")
print(classification_report(y_test, xgb_test_pred))

print("LIGHTGBM - OPTIMIZED THRESHOLD")
print(classification_report(y_test, lgbm_test_pred))
```

## Classification Threshold Optimization

The default classification threshold of 0.5 was adjusted to improve the performance of XGBoost and LightGBM.

A separate validation dataset was used to identify the threshold that maximized the F1-score.

The optimal threshold obtained was 0.5873 for XGBoost and 0.6271 for LightGBM.

After retraining both models on the complete training dataset, they were evaluated on the test dataset using their optimized thresholds.

XGBoost achieved a test precision of approximately 0.48, recall of 0.55 and F1-score of 0.51.

LightGBM achieved a test precision of approximately 0.50, recall of 0.56 and F1-score of 0.53.

LightGBM demonstrated better overall performance after threshold optimization. Therefore, LightGBM was selected as the preferred model for predicting future customer transactions.

```python
import joblib

final_model = {
    'model': lgbm_tuned,
    'threshold': lgbm_threshold,
    'features': list(X_train.columns)
}

joblib.dump(final_model, 'customer_transaction_prediction_model.pkl')

print("Final LightGBM model saved successfully!")
print("Selected threshold:", lgbm_threshold)
```

# Challenges Faced and Solutions

During the Customer Transaction Prediction project, several challenges were encountered.

**1. Imbalanced Dataset**

The dataset contained approximately 90% non-transacting customers and 10% transacting customers.

Solution: Stratified train-test splitting and class-weight balancing were used to address the class imbalance.

**2. Anonymous Features**

The dataset contained 200 anonymized variables without meaningful feature descriptions.

Solution: Feature-level exploratory analysis was limited as permitted by the project instructions. The available features were used for model training and evaluation.

**3. Model Performance Comparison**

Different algorithms produced different results, and accuracy alone did not adequately represent their performance.

Solution: Precision, recall, F1-score, ROC-AUC and confusion matrices were used to compare the models.

**4. False Positive and False Negative Predictions**

The models produced different numbers of incorrect predictions, particularly for the minority class.

Solution: Confusion matrices were examined to understand classification errors and compare model behaviour.

**5. Classification Threshold Selection**

The default threshold of 0.5 did not provide the best F1-score for the selected models.

Solution: A separate validation dataset was used to optimize the classification thresholds for XGBoost and LightGBM.

**6. Model Selection**

Selecting a model based only on accuracy could lead to an unsuitable model for identifying potential transacting customers.

Solution: Multiple evaluation metrics were considered. LightGBM was selected based on its optimized F1-score, recall, precision and ROC-AUC performance.
